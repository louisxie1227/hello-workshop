```python
def position_pnl(value_start, value_end):
  return value_end - value_start

print(position_pnl(25_000_000, 25_750_000))

def pct_return(value_start, value_end):
  return position_pnl(value_start, value_end) / value_start
print(pct_return(25_000_000, 25_750_000))
```
